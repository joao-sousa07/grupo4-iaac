```python
import pandas as pd

# 1. Carregar o dataset
df = pd.read_csv('../../data/phishing.csv')

# Separar apenas as features de entrada (remove a última coluna e colunas de ID)
X = df.iloc[:, :-1]
for col in ['Index', 'id', 'ID']:
    if col in X.columns:
        X = X.drop(columns=[col])

# 2. Calcular variância apenas das features
variances = X.var(numeric_only=True)

# 3. Filtrar atributos com variância inferior a 0.01
threshold = 0.01
low_variance_features = variances[variances < threshold]

print(f"=== ATRIBUTOS COM VARIÂNCIA QUASE NULA (< {threshold}) ===")
if not low_variance_features.empty:
    print(low_variance_features)
else:
    print("✅ Nenhum atributo de entrada apresenta variância extremamente baixa.")
```
