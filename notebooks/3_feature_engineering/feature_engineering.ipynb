```python
import pandas as pd

# 1. Carregar o dataset
df = pd.read_csv('../../data/phishing.csv')

# 2. Verificar atributos com pouca ou nenhuma variância
variances = df.var()
low_variance_cols = variances[variances < 0.01].index.tolist()

print("Atributos com variância quase nula (candidatos a remoção na Feature Engineering):")
print(low_variance_cols if low_variance_cols else "Nenhum atributo com variância extremamente baixa.")
```
