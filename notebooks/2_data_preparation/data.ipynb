```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Criar a pasta data/processed caso não exista
os.makedirs('../../data/processed', exist_ok=True)

# Carregar o dataset original
df = pd.read_csv('../../data/phishing.csv')

# Limpeza simples (remover duplicados e nulos)
df = df.drop_duplicates().dropna()

# Separar Features e Target (última coluna)
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

# Dividir em Treino (80%) e Teste (20%)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Padronizar os atributos
scaler = StandardScaler()
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns=X.columns)
X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=X.columns)

# Salvar na pasta data/processed/
X_train_scaled.to_csv('../../data/processed/X_train.csv', index=False)
X_test_scaled.to_csv('../../data/processed/X_test.csv', index=False)
y_train.to_csv('../../data/processed/y_train.csv', index=False)
y_test.to_csv('../../data/processed/y_test.csv', index=False)

print("✅ Ficheiros salvos com sucesso em data/processed/!")
```
